```python
## Pcking and Unpacking tuple
paked_tuple = 1,"hello", 3.14
print(paked_tuple)
```

```python
## Unpacking a tuple
a,b,c = packed_tuple

print(a)
print(b)
print(c)
```

```python
## unpaking with*
numbers = (1,2,3,4,5,6)
first, *middle,last = numbers
print(first)
print(middle)
print(last)
```

```python
nested_tuple = ((1,2,3), ("a", "b", "c"), (True, False))
## access the element inside a tuple
print(nested_tuple[0])
print(nested_tuple[1][2])
```

```python
## Iterating over nested tuple
for sub_tuple in nested_tuple:
    for item in sub_tuple:
        print(item,end =" ")
    print()
```
