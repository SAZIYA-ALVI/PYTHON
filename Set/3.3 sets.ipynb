```python
## create a set
my_set = {1,2,3,4,5}
print(my_set)
print(type(my_set))
```

```python
my_empty_set = set()
print(type(my_empty_set))
```

```python
my_empty_set = set([1,2,3,4,5,5,6,6,7,7,8,2,3,1])
print(my_empty_set)
```

Basic sets operation

```python
## adding and removing elements
my_set.add(11)
print(my_set)
```

```python
## Removing the element 
my_set.remove(11)
print(my_set)
```

```python
my_set.remove(4)
```

```python
my_set.discard(7)
print(my_set)
```

Pop Method

```python
removed_element = my_set.pop()
print(removed_element)
print(my_set)
```


Clear all the element

```python
my_set.clear()
print(my_set)
```

```python
## set membership test
my_set = {1,2,3,4,5}
print(3 in my_set)
print(10 in my_set)
```

```python
## Mathematical operation
set1 = {1,3,2,4,6,5}
set2 = {2,5,6,7,8,9}
## union
union_set = set1.union(set2)
print(union_set)
```

```python
## Intersection
intersection_set = set1.intersection(set2)
print(intersection_set)
```

```python
set1.intersection_update(set2)
print(set1)
```

```python
set1 = {1,2,3,4,5,6,}
set2 = {4,5,6,7,8,2}
## Diffrention
print(set1.difference(set2))
 
```

```python
## sets Methods 
set1 = {1,2,3,4,5}
set2 = {3,4,5}
## is subset
print(set1.issubset(set2))
```

```python
print(set1.issuperset(set2))
```
