```python
## accessing dictionary element
print(students['grade'])
print(students['age'])

```

```python
## accesing using get() element
print(students.get('grade'))
print(students.get('last_name'))
print(students.get('last_name', "not available"))
```

```python
## modifying dic element
## dic are mutable, update or deleted elements
print(students)
```

```python
students["age"] = 25## update value
print(students)
students["adrees"] = "india"## added new key value
print(students)
```

```python
## dictionary method
keys  = students.keys()## get all the keys

print(keys)
values = students.values()## get all value
print(values)

items = students.items() ## get  all keys value pair
print(items)
```

```python
## shallow copy
students_copy = students
print(students)
print(students_copy)
```

```python
students["name"] = "john2"
print(students)
print(students_copy)
```

```python

students_copy1 = students.copy() ## shallow copy
print(students_copy1)
print(students)
```

```python
students["name"] = "john3"
print(students_copy1)
print(students)
```

```python
## iterating over dic
## toy can use loop to iterative
for keys in students.keys():
    print(keys)
```

```python
## iterate over value
for value in students.values():
    print(value)
```

```python
## iterate over key value pairs
for key, value in students.items():
    print(f"{key}: {value}")
```

```python
## nested dic
students = {"student1":{"name":"john","age":"32"},
            "student2":{"name":"peter", "age":"23"}
             }
print(students)
```

```python
## access nested dic element
print(students["student2"]["name"])
print(students["student2"]["age"])
```

```python
## iterating over nested dic
for students_id, students_info in students.items():
    print(f"{students_id}:{students_info}")
    for key , values in students_info.items():
        
        print(f"{key}:{values}")
```

```python
## dic comherehnsion
squares = {x:x**2 for x in range(5)}
print(squares)
```

```python
## condition dic comprehension
even = {x:x**2 for x in range(10) if x%2 == 0}
print(even)
```

```python
## practical example
## use a dic to count he frequentcy of element in list
numbers = [1,2,2,3,4,5,5,4,4,4,2,2,2,1,1]
frequency = {}
for number in numbers:
    if number in frequency:
        frequency[number]+=1
    else:
        frequency[number]=1
        print(frequency)
```

```python
## merge two dic into one
dic1 = {"a":"1", "b":"2"}
dic2 = {"b":"3","c":"4"}
merged_dic = {**dic1,**dic2}
print(merged_dic)
```
