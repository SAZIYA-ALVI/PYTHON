Write a program to find the largest number in a list

```python
#✅ Method 1: Using loop (important for logic building)
number = [12,23,21,34,2,14,5,40]
largest = number[0]
for num in number:
    if num>largest:
        largest = num
print("largest number is :", largest)
```

✅ Method 2: User input version

```python
number = list(map(int, input ("enter numbers:").split()))
largest = number[0]
for num in number:
    if num>largest:
          largest = num
    print("the number is largest:", largest)
```

🧩 Problem: Find the Second Largest Number in a List

```python
numbers = [12,23,2,1,22,4,13,21,25,34,54]
largest = second_largest = float('-inf')
for num in numbers:
    if num>largest:
        second_largest= largest
        largest= num
    elif num>second_largest and num != largest:
     second_largest = num
print("second_largest number is " ,second_largest)
```

2️⃣ Count Even and Odd Numbers

```python
#Input: [1, 2, 3, 4, 5]
numbers = [1,2,3,4,5]
even = 0
odd = 0
for num in numbers:
    if num%2 == 0:
        even += 1
    else:
        odd += 1
    print("even number: ", even)
    print("odd number:" , odd)
```

3️⃣ Count frequency of elements

```python
numbers = [1,1,2,3,4,44]
freq = {}
for num in numbers:
    freq[num] = freq.get(num, 0) + 1
    print(freq)
```

Q10. Find the most frequent element in a list

```python
from collections import Counter
number= [1,1,2,2,2,3,4,7]
print(Counter(number).most_common(1))

```
