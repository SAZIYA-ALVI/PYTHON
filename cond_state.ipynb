```python
## else
## else statement execute a block of code if the condition in the if statement is false
age = 18
if age>=18:
    print("eligible for vote")
else:
    print("not eligible for vote")
```

```python
## elif statement
## allows you to check multiple condition . it stand for "else if"
age = 20
if age<13:
    print("you are a child")
elif age<18:
    print("you are teenger")
else:
    print("you are adult")
```

```python
num = int(input("Enter the number"))
if num>0:
        print("The number is positive")
        if num%2==0:
           print("The number is even")
        else:
           print("The odd number")
else:
  print("The number is zero or negative")
```

```python
## Practical examples
## determine if a year is a leap year using nested condition statements
year = int(input("Enter the year"))
if year%4 == 0:
   if year%100 == 0:
      if year%400 == 0:
           print("year, is a leap year")
      else:
             print(year, "is not a leap year")
   else:
            print(year,"is a leap year")
else:
    print(year, "is not a leap year")
         
```
